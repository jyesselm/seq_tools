# seq_tools: Introduction

Welcome to the `seq_tools` package! This package provides comprehensive tools for working with nucleic acid sequences (DNA and RNA) in Python.

## What is seq_tools?

`seq_tools` is a Python package designed for:
- **Sequence manipulation**: Convert between DNA and RNA, compute reverse complements, and more
- **Sequence analysis**: Calculate molecular weights, extinction coefficients, and analyze structures
- **Batch processing**: Work efficiently with large collections of sequences using pandas DataFrames
- **RNA structure**: Fold RNA sequences and work with secondary structures
- **Library analysis**: Calculate edit distances, generate variants, and more

## Package Overview

The `seq_tools` package provides two main interfaces:

1. **Sequence-level functions**: For working with individual sequences
2. **DataFrame-level functions**: For batch operations on multiple sequences

## Quick Start Example

Let's start with a simple example to get familiar with the package:


````python
from seq_tools import (
    to_rna,
    get_reverse_complement,
    get_molecular_weight,
    sequence_to_dataframe,
    fold,
)

# Start with a DNA sequence
dna_seq = "ATCGATCG"

# Convert to RNA
rna_seq = to_rna(dna_seq)
print(f"DNA: {dna_seq}")
print(f"RNA: {rna_seq}")

# Get reverse complement
rc = get_reverse_complement(dna_seq, "DNA")
print(f"Reverse complement: {rc}")

# Calculate molecular weight
mw = get_molecular_weight(rna_seq, "RNA")
print(f"Molecular weight (RNA): {mw:.2f} Da")

# Convert to DataFrame and fold
df = sequence_to_dataframe(rna_seq, name="example_seq")
df = fold(df)
print(f"\nFolded structure: {df.iloc[0]['structure']}")
print(f"Minimum free energy: {df.iloc[0]['mfe']:.2f} kcal/mol")
````
