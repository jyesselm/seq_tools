# Sequence-Level Operations

This notebook demonstrates how to work with individual nucleic acid sequences using `seq_tools`.

## Basic Sequence Conversions

The package provides functions to convert between DNA and RNA sequences.


```python
from seq_tools import to_dna, to_rna, to_dna_template

# DNA sequence
dna = "ATCGATCGATCG"
print(f"Original DNA: {dna}")

# Convert DNA to RNA (T -> U)
rna = to_rna(dna)
print(f"Converted to RNA: {rna}")

# Convert RNA back to DNA (U -> T)
dna_again = to_dna(rna)
print(f"Converted back to DNA: {dna_again}")

# Convert to DNA template (includes T7 promoter)
template = to_dna_template(rna)
print(f"DNA template with T7 promoter: {template}")
```

## Reverse Complement

The reverse complement is essential in molecular biology. It represents the complementary strand read in the reverse direction.


```python
from seq_tools import get_reverse_complement

# DNA reverse complement
dna_seq = "ATCGATCG"
rc_dna = get_reverse_complement(dna_seq, "DNA")
print(f"DNA sequence:        {dna_seq}")
print(f"Reverse complement:  {rc_dna}")

# RNA reverse complement
rna_seq = "AUCGAUCG"
rc_rna = get_reverse_complement(rna_seq, "RNA")
print(f"\nRNA sequence:        {rna_seq}")
print(f"Reverse complement:  {rc_rna}")

# Verify: reverse complement of reverse complement should be original
print(f"\nVerification (DNA): {get_reverse_complement(rc_dna, 'DNA') == dna_seq}")
print(f"Verification (RNA): {get_reverse_complement(rc_rna, 'RNA') == rna_seq}")
```

## Molecular Weight Calculation

Calculate the molecular weight of nucleic acid sequences. This is useful for:
- Determining concentrations
- Oligo synthesis calculations
- Mass spectrometry analysis


```python
from seq_tools import get_molecular_weight

# Single-stranded DNA
dna_seq = "ATCGATCG"
mw_dna_ss = get_molecular_weight(dna_seq, "DNA", double_stranded=False)
print(f"Single-stranded DNA '{dna_seq}': {mw_dna_ss:.2f} Da")

# Double-stranded DNA (includes both strands)
mw_dna_ds = get_molecular_weight(dna_seq, "DNA", double_stranded=True)
print(f"Double-stranded DNA '{dna_seq}': {mw_dna_ds:.2f} Da")

# Single-stranded RNA
rna_seq = "AUCGAUCG"
mw_rna_ss = get_molecular_weight(rna_seq, "RNA", double_stranded=False)
print(f"Single-stranded RNA '{rna_seq}': {mw_rna_ss:.2f} Da")

# Compare DNA vs RNA
print(f"\nDifference (RNA - DNA): {mw_rna_ss - mw_dna_ss:.2f} Da")
```

## Maximum Stretch

Find the longest stretch of identical nucleotides in a sequence. This can be useful for identifying homopolymer regions.


```python
from seq_tools import get_max_stretch

sequences = ["AAATTTGGGCCC", "ATCGATCGATCG", "AAAAATTTTTGGGGG", "GGGGUUUUCCCCAAAA"]

for seq in sequences:
    max_stretch = get_max_stretch(seq)
    print(f"Sequence: {seq:20} | Max stretch: {max_stretch}")
```

## Extinction Coefficient

The extinction coefficient (ε) is used to calculate nucleic acid concentration from UV absorbance measurements. It's measured in M⁻¹cm⁻¹.


```python
from seq_tools import get_extinction_coeff

# DNA extinction coefficient
dna_seq = "ATCGATCGATCG"
ec_dna = get_extinction_coeff(dna_seq, "DNA", double_stranded=False)
print(f"DNA sequence: {dna_seq}")
print(f"Extinction coefficient (single-stranded): {ec_dna:,} M⁻¹cm⁻¹")

ec_dna_ds = get_extinction_coeff(dna_seq, "DNA", double_stranded=True)
print(f"Extinction coefficient (double-stranded): {ec_dna_ds:,} M⁻¹cm⁻¹")

# RNA extinction coefficient
rna_seq = "AUCGAUCGAUCG"
ec_rna = get_extinction_coeff(rna_seq, "RNA", double_stranded=False)
print(f"\nRNA sequence: {rna_seq}")
print(f"Extinction coefficient: {ec_rna:,} M⁻¹cm⁻¹")

# Calculate concentration from absorbance
# Concentration (M) = A260 / (ε × path_length)
# Example: A260 = 0.5, path_length = 1 cm
A260 = 0.5
path_length = 1.0
concentration = A260 / (ec_dna * path_length) * 1e6  # Convert to μM
print(f"\nExample: For A260 = {A260}, concentration = {concentration:.2f} μM")
```
